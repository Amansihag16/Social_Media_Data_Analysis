# SocialPulse – Exploratory Data Analysis & Analytics Validation
**An End-to-End Big Data Engineering Solution for Social Media Sentiment & Engagement Analytics**

This notebook demonstrates exploratory analysis, data quality checks, and sentiment visualization across the Bronze, Silver, and Gold layers.

```python
import sys
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set project paths
PROJECT_ROOT = Path("..")
GOLD_DIR = PROJECT_ROOT / "data" / "processed" / "gold"
SILVER_DIR = PROJECT_ROOT / "data" / "delta" / "silver"

sns.set_theme(style="whitegrid")
print("Environment initialized successfully.")
```

## 1. Inspecting Silver Transformed & Enriched Posts

```python
posts_df = pd.read_parquet(SILVER_DIR / "silver_social_posts" / "data.parquet")
print(f"Loaded {len(posts_df)} enriched social media posts across {posts_df['source_platform'].nunique()} platforms.")
posts_df[["source_platform", "author", "sentiment_label", "sentiment_compound", "engagement_score", "cleaned_text"]].head(5)
```

## 2. Gold Sentiment Summary & Distribution

```python
sent_summary = pd.read_csv(GOLD_DIR / "gold_sentiment_summary.csv")
display(sent_summary)

# Plot Sentiment Distribution across Platforms
plat_sent = sent_summary[sent_summary["platform"] != "ALL_PLATFORMS"]
fig, ax = plt.subplots(figsize=(8, 4))
plat_sent.set_index("platform")[["positive_posts", "neutral_posts", "negative_posts"]].plot(
    kind="bar",
    stacked=True,
    color=["#10B981", "#6B7280", "#EF4444"],
    ax=ax
)
plt.title("Sentiment Volume Distribution by Platform", fontsize=13, fontweight="bold")
plt.ylabel("Post Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 3. Platform Performance & Engagement Metrics

```python
perf_df = pd.read_csv(GOLD_DIR / "gold_platform_performance.csv")
display(perf_df)

plt.figure(figsize=(7, 4))
sns.barplot(data=perf_df, x="platform", y="avg_engagement_per_post", palette="Blues_d")
plt.title("Average Engagement Per Post by Platform", fontsize=13, fontweight="bold")
plt.ylabel("Avg Interactions (Likes + Comments + Shares)")
plt.tight_layout()
plt.show()
```

## 4. Top Trending Hashtags

```python
tags_df = pd.read_csv(GOLD_DIR / "gold_hashtag_trends.csv")
top_tags = tags_df.groupby("hashtag")["usage_frequency"].sum().reset_index().sort_values(by="usage_frequency", ascending=False).head(10)

plt.figure(figsize=(9, 4))
sns.barplot(data=top_tags, y="hashtag", x="usage_frequency", color="#3B82F6")
plt.title("Top 10 Trending Hashtags Across Multi-Source Feeds", fontsize=13, fontweight="bold")
plt.xlabel("Frequency")
plt.tight_layout()
plt.show()
```

## 5. Customer Concerns & Complaint Diagnostics

```python
concerns_df = pd.read_csv(GOLD_DIR / "gold_customer_concerns.csv")
top_concerns = concerns_df.groupby("issue_category")["frequency"].sum().reset_index().sort_values(by="frequency", ascending=False)

plt.figure(figsize=(8, 4))
sns.barplot(data=top_concerns, x="frequency", y="issue_category", color="#EF4444")
plt.title("Customer Concerns by Problem Taxonomy", fontsize=13, fontweight="bold")
plt.xlabel("Total Frequency Mentioned")
plt.tight_layout()
plt.show()
```
